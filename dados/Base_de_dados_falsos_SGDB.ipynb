```python
from faker import Faker
import unicodedata
import random
import re
import pandas as pd

fake = Faker('pt_BR')
random.seed(42)
Faker.seed(42)

DOMINIOS = [
    "gmail.com", "hotmail.com", "outlook.com",
    "yahoo.com.br", "icloud.com", "uol.com.br"
]

PREFIXOS_TRATAMENTO = re.compile(
    r'^(sr\.?|sra\.?|dr\.?|dra\.?|prof\.?|profa\.?)\s+',
    flags=re.IGNORECASE
)

def normalizar(texto):
    nfkd = unicodedata.normalize('NFKD', texto)
    return ''.join(c for c in nfkd if not unicodedata.combining(c))

def email_do_nome(nome):
    nome_limpo = PREFIXOS_TRATAMENTO.sub('', nome).strip()
    partes = nome_limpo.split()
    primeiro  = normalizar(partes[0]).lower()
    sobrenome = normalizar(partes[-1]).lower() if len(partes) > 1 else ""
    estilo = random.choice(["ponto", "underscore", "junto", "abrev"])
    numero = random.choice(["", str(random.randint(1, 99)), str(random.randint(1900, 2005))])
    dominio = random.choice(DOMINIOS)
    if estilo == "ponto" and sobrenome:
        usuario = f"{primeiro}.{sobrenome}"
    elif estilo == "underscore" and sobrenome:
        usuario = f"{primeiro}_{sobrenome}"
    elif estilo == "junto" and sobrenome:
        usuario = f"{primeiro}{sobrenome}"
    else:
        inicial = sobrenome[0] if sobrenome else ""
        usuario = f"{primeiro}{inicial}"
    usuario = re.sub(r'[^a-z0-9._]', '', usuario + numero)
    return f"{usuario}@{dominio}"


quantidade = 20000
dados = []

for i in range(quantidade):
    nome = fake.name()
    dados.append({
        "id":         i + 1,
        "nome":       nome,
        "email":      email_do_nome(nome),
        "telefone":   fake.phone_number(),
        "celular":    fake.cellphone_number(),
        "logradouro": fake.street_name(),
        "numero":     fake.building_number(),
        "bairro":     fake.bairro(),
        "cidade":     fake.city(),
        "estado":     fake.estado_sigla(),
        "cep":        fake.postcode(),
    })

df = pd.DataFrame(dados)
df.to_excel("dados_falsos.xlsx", index=False)

print(f"✅ {len(df)} registros exportados para dados_falsos.xlsx")
print(df[["nome", "email", "cidade", "estado"]].to_string(index=False))
```
